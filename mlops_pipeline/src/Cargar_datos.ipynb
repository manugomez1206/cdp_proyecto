```python
# Importamos librerías necesarias
import pandas as pd
import os
import io
from google.cloud import storage
```

```python
# Lectura de datos desde Google Cloud Storage
BUCKET_NAME = "cdp-proyecto-datos"
FILE_NAME   = "Base_de_datos.csv"

def leer_datos_gcp(bucket_name, file_name):
    """
    Lee el dataset desde Google Cloud Storage
    y retorna un DataFrame de pandas.
    """
    client  = storage.Client()
    bucket  = client.bucket(bucket_name)
    blob    = bucket.blob(file_name)
    data    = blob.download_as_bytes()
    df      = pd.read_csv(io.BytesIO(data))
    print(f"Datos cargados desde GCP")
    print(f"Shape: {df.shape[0]} filas x {df.shape[1]} columnas")
    return df
```

```python
# Cargar datos desde GCP
df = leer_datos_gcp(BUCKET_NAME, FILE_NAME)

# Tipos de datos y nulos
print("=== TIPOS DE DATOS ===")
print(df.dtypes)

print("\n=== VALORES NULOS ===")
print(df.isnull().sum())

print("\n=== MUESTRA ===")
df.head()
```

```python
# Guardar como Excel para el pipeline
BASE_DIR    = os.path.abspath(os.path.join(os.getcwd(), ".."))
OUTPUT_PATH = os.path.join(BASE_DIR, "Base_de_datos.xlsx")
df.to_excel(OUTPUT_PATH, index=False)

print(f"Archivo guardado en: {OUTPUT_PATH}")
```
